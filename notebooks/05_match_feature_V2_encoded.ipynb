```python
import pandas as pd
from pathlib import Path
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

PROJECT_ROOT = Path.cwd().parent

RAW_DATA = PROJECT_ROOT / "FIFA WC Dataset"
CLEANED_DATA = PROJECT_ROOT / "cleaned_data"
PROCESSED_DATA = PROJECT_ROOT / "processed_data"
FEATURE_DATA = PROJECT_ROOT / "feature_data"
ML_DATA = PROJECT_ROOT / "ml_data"
ML_DATA.mkdir(exist_ok=True)

#results = pd.read_csv(CLEANED_DATA / "results_2000_onwards.csv")
#team_matches = pd.read_csv(PROCESSED_DATA / "team_matches.csv")
match_feature = pd.read_csv(FEATURE_DATA / "match_feature_V2.csv")
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
print(df["date"].min())
print(df["date"].max())
```

```python
# going to create Train and Test df.
train_df = df[df["date"] < "2024-01-01"].copy()
test_df = df[df["date"] >= "2024-01-01"].copy()
```

```python
train_df.tail(3)
```

```python
test_df.head(3)
```

```python
# checked the shape of both, accurate.
print(train_df.shape)
print(test_df.shape)
```

```python
train_df = train_df.drop(columns = ["date"])
test_df = test_df.drop(columns = ["date"])
```

```python
X_train = train_df.drop(columns=["target"])
y_train = train_df["target"]

X_test = test_df.drop(columns=["target"])
y_test = test_df["target"]
```

```python
print(X_train.shape)
print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

```python
train_encoded = encoder.fit_transform(X_train[["tournament"]])
```

```python
test_encoded = encoder.transform(X_test[["tournament"]])
```

```python
encoded_columns = encoder.get_feature_names_out(["tournament"])
```

```python
train_encoded = pd.DataFrame(
    train_encoded,
    columns=encoded_columns,
    index=X_train.index
)

test_encoded = pd.DataFrame(
    test_encoded,
    columns=encoded_columns,
    index=X_test.index
)
```

```python
X_train = X_train.drop(columns=["tournament"])
X_test = X_test.drop(columns=["tournament"])
```

```python
X_train = pd.concat([X_train, train_encoded], axis=1)
X_test = pd.concat([X_test, test_encoded], axis=1)
```

```python
print(X_train.info())
print(X_test.info())
```

```python
print(X_train.isnull().sum().sum())
print(X_test.isnull().sum().sum())
print(X_train.shape)
print(X_test.shape)
```

```python
# Reattaching the target columns to DFs.
train_final = X_train.copy()
train_final["target"] = y_train

test_final = X_test.copy()
test_final["target"] = y_test
```

```python
print(train_final.shape)
print(test_final.shape)
```

```python
train_final.to_csv(ML_DATA / "train.csv", index=False)
test_final.to_csv(ML_DATA / "test.csv", index=False)
```
