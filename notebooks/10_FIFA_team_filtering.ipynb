```python
results["home_team"] = results["home_team"].replace(team_mapping)
results["away_team"] = results["away_team"].replace(team_mapping)
```

```python
all_teams_updated = sorted(
    set(results["home_team"]).union(results["away_team"])
)
```

```python
dataset_set = set(all_teams_updated)
fifa_set = set(fifa_teams["team"])

not_in_fifa = sorted(dataset_set - fifa_set)
missing_from_dataset = sorted(fifa_set - dataset_set)

print(f"Teams not in FIFA list: {len(not_in_fifa)}")
print(f"FIFA teams missing from dataset: {len(missing_from_dataset)}")
```

```python
# creating Fifa team set.
fifa_team_set = set(fifa_teams["team"])
```

```python
# Filtering the matches.

results_fifa = results[
    results["home_team"].isin(fifa_team_set) &
    results["away_team"].isin(fifa_team_set)
].copy()
```

```python
# checking original, filtered and removed matches. 

print(f"Original matches : {len(results):,}")
print(f"Filtered matches : {len(results_fifa):,}")
print(f"Matches removed  : {len(results) - len(results_fifa):,}")
```

```python
filtered_teams = sorted(
    set(results_fifa["home_team"]).union(results_fifa["away_team"])
)

print(f"Unique FIFA teams in filtered dataset: {len(filtered_teams)}")
```

```python
print(f"Home teams : {results_fifa['home_team'].nunique()}")
print(f"Away teams : {results_fifa['away_team'].nunique()}")
print(f"Total unique teams : {len(filtered_teams)}")
```

## Notebook Summary

This notebook standardized team names using the official FIFA member association list and removed matches involving non-FIFA teams.

### Results
- Original dataset: **25,406 matches**, **321 unique teams**
- Filtered dataset: **23,760 matches**, **210 FIFA teams**
- Matches removed: **1,646**

A total of **18 team name aliases** were standardized to match the official FIFA naming convention. The only FIFA member association absent from the final dataset is **Chinese Taipei**, as no matches involving the team were present in the source data (2000 onwards).

The resulting dataset, `results_2000_fifa211.csv`, now serves as the new source dataset for all subsequent stages of the project, including team statistics, Elo ratings, feature engineering, EDA, encoding, and machine learning.
