```python
import pandas as pd
from pathlib import Path

# Loading CSVs using path.cwd
PROJECT_ROOT = Path.cwd().parent
RAW_DATA = PROJECT_ROOT / "FIFA WC Dataset"
FEATURE_DATA = PROJECT_ROOT / "feature_data"

match_feature = pd.read_csv(FEATURE_DATA / "match_feature_V2.csv")
```

```python
# Just verifying if the dataset is correct

print(match_feature.shape)
match_feature.head()
match_feature.tail()
```

```python
# Adding 'abs_elo_diff' feature:

match_feature["abs_elo_diff"] = match_feature["elo_diff"].abs()
```

```python
# Adding 'win_pct_gap' feature:

match_feature["win_pct_gap"] = (
    match_feature["home_win_pct"] - match_feature["away_win_pct"]
).abs()
```

```python
# Adding 'last5_ppg_gap' feature:

match_feature["last5_ppg_gap"] = (
    match_feature["home_last5_ppg"] -
    match_feature["away_last5_ppg"]
).abs()
```

```python
# Adding 'avg_gf_gap' feature:

match_feature["avg_gf_gap"] = (
    match_feature["home_avg_gf"] -
    match_feature["away_avg_gf"]
).abs()
```

```python
# Adding 'avg_ga_gap' feature:

match_feature["avg_ga_gap"] = (
    match_feature["home_avg_ga"] -
    match_feature["away_avg_ga"]
).abs()
```

```python
# Checking all the added features:

new_features = [
    "abs_elo_diff",
    "win_pct_gap",
    "last5_ppg_gap",
    "avg_gf_gap",
    "avg_ga_gap"
]

match_feature[new_features].describe()
```
